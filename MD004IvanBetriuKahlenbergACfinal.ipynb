```r
# Convertir data_ais_num a formato largo
data_long <- data_ais_num %>%
  pivot_longer(cols = everything(), names_to = "variable", values_to = "value")

# Verificar que la columna 'variable' existe en data_long
if (!"variable" %in% colnames(data_long)) {
  stop("La columna 'variable' no está en data_long. Verifica la transformación a formato largo.")
}

# Obtener los nombres de las variables en data_long
var_data <- unique(data_long$variable)

# Crear una lista vacía para almacenar los gráficos de boxplot
box_plots <- list()

# Bucle para generar un boxplot por cada variable numérica en el conjunto de datos
for (variable in var_data) {

  # Filtrar los datos para la variable actual
  data_filtered <- data_long[data_long$variable == variable, ]

  # Verificar si hay datos suficientes para graficar
  if (nrow(data_filtered) > 0) {
  
    # Crear el boxplot usando ggplot2
    box_plot <- ggplot(data_filtered, aes(x = "", y = value)) +
      geom_boxplot(
        fill = "blue",        # Color de relleno
        color = "#3C0023",    # Color del borde
        na.rm = TRUE          # Ignorar valores NA
      ) +
      coord_flip() +          # Girar el boxplot horizontalmente
      labs(
        title = paste("Boxplot", variable), 
        y = variable
      ) +
      theme_minimal() +
      theme(
        text = element_text(size = 12),
        panel.background = element_rect(fill = "lightgray"),
        panel.grid.major = element_line(color = "white"),
        panel.grid.minor = element_line(color = "white")
      )
    
    # Agregar el gráfico a la lista con el nombre de la variable como clave
    box_plots[[variable]] <- box_plot
  }
}

# Configurar el tamaño de los gráficos para la representación final
options(
  repr.plot.width = 20,  
  repr.plot.height = 10   
)

# Mostrar los boxplots en una cuadrícula con 4 columnas si la lista no está vacía
if (length(box_plots) > 0) {
  grid.arrange(
    grobs = box_plots,  
    ncol = min(4, length(box_plots))  # Ajuste dinámico del número de columnas
  )
}
```

```r
# Convertir variables binarias a numéricas (0 y 1)
df <- df %>%
  mutate(across(c(is_referral, given_permission_1, given_permission_2), 
                ~ as.numeric(.) - 1))

# Verificar la transformación
str(df)
```

```r
# Filtrar solo las variables numéricas
df_numeric <- df %>%
  select(where(is.numeric))

# Verificar la estructura del nuevo dataframe
str(df_numeric)
```

```r
pca = df_numeric

# Aplicar PCA con estandarización
pca <- prcomp(pca, scale = TRUE)

# Mostrar resumen del PCA
summary(pca)
```

```r
# Gráfico de la varianza explicada
explained_var = pca$sdev^2 / sum(pca$sdev^2) * 100

explained_var_df <- data.frame(
  PC = paste0("PC", 1:length(explained_var)),
  Variance = explained_var
)

options(repr.plot.width = 10,  
  repr.plot.height = 10)

# Crear un gráfico de barras para la varianza explicada
ggplot(explained_var_df, aes(x = reorder(PC, -Variance), y = Variance)) +
  geom_bar(stat = "identity", fill = "#990000", color = "#990000") +
  geom_text(aes(label = round(Variance, 1)), vjust = -0.5, size = 5) +
  labs(
    title = "Porcentaje de Varianza Explicada por Componente Principal",
    x = "Componentes Principales",
    y = "Porcentaje de Varianza Explicada"
  ) +
  theme_minimal(base_size = 14)
```

Se decide optar por el uso de 8 factores que representan el 81% de la varianza.

```r
# Mostrar las cargas de los primeros ocho componentes principales
cargas <- as.data.frame(round(pca$rotation[, 1:8], 3))
print(cargas)
```

```r
library(FSelectorRcpp)
library(ggplot2)
library(patchwork)  # Necesario para wrap_plots()
library(purrr)  # Necesario para pmap()

# Convertir is_churned a factor para evitar warnings
df$is_churned <- factor(df$is_churned, levels = c(0,1))

# Generar una paleta de colores en la misma tonalidad (rojo)
colores <- scales::seq_gradient_pal("#ffcccc", "#990000", "Lab")(seq(0, 1, length.out = 3))

# Función para calcular y graficar la importancia de atributos
graficar_importancia <- function(data, formula, type, titulo, color) {
  ig <- information_gain(formula = formula, data = data, type = type)
  
  ggplot(ig, aes(x = reorder(attributes, importance), y = importance)) +
    geom_bar(stat = "identity", fill = color, color = color) +
    geom_text(aes(label = round(importance, 3)), color = "black", size = 5, hjust = -0.2) +
    coord_flip() +
    labs(
      title = titulo,
      x = "Atributos",
      y = "Importancia"
    ) +
    theme_minimal(base_size = 14) +
    theme(axis.text.y = element_text(size = 12))
}

# Generar gráficos para los tres tipos de métricas
tipos <- c("infogain", "gainratio", "symuncert")
titulos <- c("Importancia de Atributos (Information Gain)",
             "Importancia de Atributos (Gain Ratio)",
             "Importancia de Atributos (Symmetrical Uncertainty)")

# Usar purrr::pmap para pasar los colores a la función
graficos <- pmap(
  list(tipos, titulos, colores),
  ~ graficar_importancia(df, is_churned ~ ., ..1, ..2, ..3)
)

# Mostrar los gráficos en un layout usando patchwork
wrap_plots(graficos, ncol = 1)
```

El análisis de Information Gain Analysis (IGA) nos permite identificar las variables más relevantes para predecir el churn en los clientes. En este caso, los resultados muestran que las variables relacionadas con las transacciones y el uso de recompensas tienen el mayor impacto en la predicción.

Las tres métricas analizadas (Information Gain, Gain Ratio y Symmetrical Uncertainty) coinciden en destacar como más importantes las siguientes variables:

payments_completed_amount_first_7days,
reward_purchase_count_first_7days,
first_payment_amount,
payments_completed,
coins_redeemed_first_7days.
Esto sugiere que el comportamiento de pago y el uso de recompensas dentro de los primeros días de actividad del cliente son factores clave para determinar si hará churn o no.

Por otro lado, variables como visits_feature_2, payments_failed, device y age aparecen con una importancia muy baja en todas las métricas, lo que indica que tienen poca influencia en la predicción y podrían descartarse para optimizar el modelo.

# Desarrollo del modelo de regresión logística

## MODELO 1(Basado en PCA)

```r
# Selección variables modelo basado en pca
df_subset = df_final 
# Convertir variable objetivo en factor
df_subset$is_churned = factor(df_subset$is_churned,levels = c('0','1'))
```

```r
# Cargar el paquete
library(rsample)
set.seed(50)

# Partición de los datos en entrenamiento y prueba
split = initial_split(df_subset, prop = 0.8, strata = is_churned)
df_train = training(split)
df_test = testing(split)

# Definir las variables objetivo 
ytrain = df_train$is_churned
ytest = df_test$is_churned
```

## TRAIN

```r
# Verificar si is_churned está correctamente definido
df_train$is_churned <- factor(df_train$is_churned, levels = c(0,1))

# Ajustar la regresión logística usando todas las variables excepto is_churned
mrl1 <- glm(is_churned ~ ., data = df_train, family = "binomial")

# Mostrar resultados
summary(mrl1)
```

```r
# Cargar el paquete
library(caret)

# Predicciones de probabilidades
mrl1_prob <- predict(mrl1, df_train, type = "response")

# Predicción binaria
mrl1_pred <- ifelse(mrl1_prob > 0.5, 1, 0)
# Matriz de confusión
confusionMatrix(as.factor(mrl1_pred), as.factor(ytrain))
```

El modelo presenta un accuracy del 72.42%, pero con una especificidad muy baja (7.76%), lo que indica que tiene dificultades para detectar correctamente los casos de churn (clase 1). La sensibilidad es alta (98.43%), lo que significa que el modelo predice correctamente la mayoría de los casos de no churn (clase 0). Sin embargo, el bajo valor de especificidad sugiere que el modelo aún presenta un problema de desbalance y no logra distinguir correctamente entre clientes que hacen churn y los que no. Para mejorar, se opta por ajustar el umbral de decisión.

```r
# Probar con un umbral más bajo
mrl1_pred <- ifelse(mrl1_prob > 0.3, 1, 0)

# Matriz de confusión
confusionMatrix(as.factor(mrl1_pred), as.factor(ytrain))
```

Después de aplicar un umbral de 0.3, el modelo mejora en la detección de la clase 1 (churn), logrando una sensibilidad del 62.97%, lo que indica que ahora es capaz de identificar más clientes que hacen churn en comparación con el modelo anterior. Además, la especificidad es del 61.42%, lo que significa que también ha mejorado en la identificación de clientes que no hacen churn.

El accuracy del modelo es del 62.53%, lo que muestra una ligera disminución respecto a la versión anterior, pero se ha logrado un mejor equilibrio en la detección de ambas clases. 
En general, ajustar el umbral ha permitido mejorar la capacidad del modelo para detectar churn, aunque su capacidad predictiva sigue siendo limitada.

```r
# HACER CURVA ROC
# Instalar y cargar pROC si no lo tienes instalado
if (!require("pROC")) install.packages("pROC", dependencies=TRUE)
library(pROC)

# Calcular la curva ROC
roc_curve <- roc(ytrain, mrl1_prob)

# Plotear la curva ROC
plot(roc_curve, col="blue", lwd=2, main="Curva ROC - Modelo de Regresión Logística")

# Calcular y mostrar el AUC
auc_value <- auc(roc_curve)
cat("Área bajo la curva (AUC):", auc_value, "\n")
```

La curva ROC evalúa el desempeño del modelo de clasificación binaria para distinguir entre clientes que hacen churn (1) y los que no (0). El AUC de 0.668 indica que el modelo tiene un rendimiento superior al azar, pero sigue siendo reducido y lejos de un clasificador óptimo (AUC cercano a 1).

El modelo identifica correctamente a los clientes que no hacen churn (especificidad del 78.93%), pero tiene dificultades para detectar a aquellos que sí lo hacen (sensibilidad del 48.87%). Como resultado la precisión del modelo se reduce al 57%.

El AUC de 0.738 indica una mejora en la capacidad del modelo para diferenciar entre clientes que hacen churn y los que no.
